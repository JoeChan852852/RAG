# 📚 1. Introduction: How Does a RAG Chatbot Work?

Before we start building, let’s understand the basic idea behind a RAG chatbot.

---

## 1.1 🤖 What Is a “Normal” Chatbot?

A normal AI chatbot, like ChatGPT, can answer many questions because it has learned from a large amount of text.

But it has some limits:

- It may not know about **new or recent information**
- It cannot automatically check your **class notes, PDFs, or private documents**
- It may give an answer even when it is not sure
- Sometimes, it may **hallucinate**

💡 **Hallucinate** means the AI gives an answer that sounds confident, but is actually wrong.

Think of a normal chatbot like a student who studied a lot before the exam, but is **not allowed to open any notes during the exam**.

---

## 1.2 🔍 What Is a RAG Chatbot?

**RAG** stands for **Retrieval-Augmented Generation**.

A RAG chatbot is like giving the AI an **open-book exam**.

Instead of only using what it already “remembers,” the chatbot first looks inside your documents, finds useful information, and then writes an answer based on that information.

---

## 1.3 🧠 How RAG Works

A RAG chatbot follows three simple steps:

1. **Retrieve**  
   The chatbot searches your documents and finds the most relevant parts.

2. **Augment**  
   It adds those document parts to your question as extra context.

3. **Generate**  
   The AI writes an answer using your question and the retrieved information.

---

A normal chatbot answers from memory.

A RAG chatbot answers by first checking your documents.

That makes it more useful when you want answers based on:

*   Lecture notes
*   Research papers
*   Company documents
*   Course materials
*   PDFs or reports

# 4. 🧱 The Naive Solution: Prompt Stuffing

## 🧪 Experiment 2: Giving the AI Everything at Once

Before learning RAG, let’s try the simplest idea:

> “What if we just copy and paste the whole document into the prompt?”

This approach is called **prompt stuffing**.

---

### 4.1 📌 What Is Prompt Stuffing?

Prompt stuffing means putting a large amount of text directly into the chat box and asking the AI to answer based on it.

For example:

1. Copy a long document  
2. Paste it into the prompt  
3. Add your question at the end  
4. Ask the AI to answer  

It sounds simple — but it does not work well for large documents.

---

### 4.2 Read a PDF
```mermaid
flowchart LR
    Start[Start] --> Open[Open PDF]
    Open --> Process[Process each page<br/>- Extract text<br/>- Add page header<br/>- Add text or 'No text']
    Process --> More{More pages?}
    More -->|Yes| Process
    More -->|No| End[Extraction Complete]
```

`````python
import fitz  # PyMuPDF: The library used to "read" and parse PDF files

# The file path is the specific location of your technical manual
pdf_path = "F18-ABCD-000.pdf"

# --- 1. ACCESS THE DOCUMENT ---
# Open the file so we can begin scanning its contents
doc = fitz.open(pdf_path)
print(f"Success! Opened a PDF with {len(doc)} pages\n")

# --- 2. INITIALIZE THE "COLLECTION" ---
# This variable will hold all the text we extract from every page.
# Think of it as a single, long digital transcript.
PDF_content = ""

# --- 3. THE EXTRACTION LOOP ---
# We visit every page one-by-one, from the first to the last.
for i in range(len(doc)):
    page = doc[i]  # "Pick up" the current page
    
    # OCR/Text Scanning: Convert the visual layout into readable strings.
    text = page.get_text("text") 

    # --- 4. FORMATTING THE TRANSCRIPT ---
    # We add visual dividers (═) so the AI (and we) can see where pages start and end.
    # This acts like a "table of contents" built directly into the text.
    PDF_content += f"\n{'═' * 80}\n"
    PDF_content += f"Page {i+1:3d} of {len(doc)}\n"
    PDF_content += f"{'═' * 80}\n\n"

    # --- 5. VALIDATION ---
    # We check if the page actually contains text.
    # If a page is just a photo or diagram, the "text" will be empty.
    if text.strip():
        PDF_content += text + "\n\n"
    else:
        PDF_content += "[No extractable text on this page - likely an image or diagram]\n\n"

# --- 6. FINALIZING ---
# Close the file to free up system memory.
doc.close()

# The 'PDF_content' variable is now ready to be sent to an AI or a database!
print("Extraction complete. Text is ready for processing.")
`````

## 4.4 ⚠️ Why Prompt Stuffing Is a Problem

![My Image](images/photo2.png)


If your document is very long, such as a **1,000-page manual**, prompt stuffing does not work well.

There are **three** main problems:

---

### 4.4.1. The Chat Box Has a Limit

![My Image](images/photo3.png)

AI models can only handle a limited amount of text at one time.

This limit is called the **context window**.

The context window includes everything in the request, such as:

- System instructions
- Chat history
- Your question
- The document text you paste in
- The AI’s final answer

In simple terms:

> **Input tokens + output tokens must fit inside the model’s limit.**

If the request is too large, the API may return an error such as:

```text
maximum context length exceeded
````

or

```text
context_length_exceeded
```

This means the prompt is too big for the model to process.

***

### 4.4.2. It Can Be Slow and Expensive

![My Image](images/photo4.png)

The more text you send to the AI, the more tokens it needs to process.

This can make your chatbot:

*   Slower to respond
*   More expensive to run
*   Harder to scale for many users

So even if the model can accept a large prompt, sending everything is usually not efficient.

***

### 4.4.3. The AI Can Get “Lost in the Middle”

![My Image](images/photo5.png)

Even when a model supports a large context window, it may not use all parts of the prompt equally well.

Research shows that LLMs often perform better when important information is near the:

*   **Beginning** of the prompt
*   **End** of the prompt

But they may perform worse when the key information is hidden in the **middle** of a long prompt.

This problem is called:

> **Lost in the Middle**

So if you paste a huge document into the prompt, the AI might still miss the most important detail.

Sources:

*   <https://arxiv.org/abs/2307.03172>
*   <https://direct.mit.edu/tacl/article/doi/10.1162/tacl_a_00638/119630/Lost-in-the-Middle-How-Language-Models-Use-Long>

***

## 💡 Key Takeaway

![My Image](images/photo7.png)

Prompt stuffing is simple, but it is not a good solution for large documents.

Instead of giving the AI **everything**, we need a smarter method.

# 4. The Final Solution: Retrieval‑Augmented Generation (RAG)

## 🏗️ The 6-Stage RAG Pipeline

RAG is the **smart way** to answer questions from large documents.

Instead of sending the whole book to the AI, we only send the **most relevant parts**.

---

## 4.1 🔄 How RAG Works

RAG follows 6 simple stages:

### 4.1.1. 📥 Load

Read the document, such as:

- PDF
- Text file
- Word document
- Website content

**Goal:** Get the document into the system.

---

### 4.1.2. ✂️ Split

Break the long document into smaller pieces called **chunks**.

For example, one long PDF may be split into many small paragraphs.

**Goal:** Make the document easier to search.

---

### 4.1.3. 🔢 Embed

Turn each chunk of text into numbers.

These numbers are called **vectors**.

A vector represents the **meaning** of the text.

**Goal:** Help the computer understand and compare meaning.

---

### 4.1.4. 🗄️ Store

Save the vectors in a special database called a **Vector Database**.

**Goal:** Store the document chunks so we can search them later.

---

### 4.1.5. 🔍 Retrieve

When the user asks a question, the system searches the Vector Database.

It finds the chunks that are most related to the question.

**Goal:** Get only the useful information.

---

### 4.1.6. ✍️ Generate

The AI uses:

- The user’s question
- The retrieved document chunks

Then it writes the final answer.

**Goal:** Give an answer based on the document, not just the AI’s memory.

---

## 4.2 🧠 Simple Summary

```text
Load → Split → Embed → Store → Retrieve → Generate
````

In one sentence:

> RAG reads your documents, finds the most relevant parts, and gives them to the AI to write a better answer.

***

## 4.3 🔢 Why Do We Need Vectors?

Computers do not understand text like humans do.

So we turn text into **vectors**, which are lists of numbers that represent meaning.

This allows the computer to compare ideas, not just exact words.

***

## 4.4 🔎 Example: Normal Search vs Vector Search

Imagine you search for:

> “Jet Engine”

### 4.4.1 Normal Search

A normal search looks for the exact words:

> “Jet” and “Engine”

It may miss related words like:

* Turbine
* Propulsion
* Aircraft engine

This is because normal search is usually based on **keyword matching**.

For example, a document may be scored like this:


$$ \text{Score} = \text{number of matching keywords} $$


If your query is:

$$ Q = \{\text{Jet}, \text{Engine}\} $$


And a document contains:

$$ D_1 = \{\text{Jet}, \text{Engine}\} $$

Then the score is high because both words match:


$$ \text{Score}(Q, D_1) = 2 $$


But if another document contains:

$$ D_2 = \{\text{Turbine}, \text{Thrust}, \text{Aircraft}\} $$

Then normal search may give it a low score:


$$ \text{Score}(Q, D_2) = 0 $$

Even though the document is clearly related to **jet engines**, the exact words do not appear.

***

### 4.4.2 Vector Search

A vector search understands meaning.

So it can find related ideas, such as:

* “Turbine”
* “Propulsion system”
* “Aircraft engine”
* “Thrust”

Even if the exact words **“Jet Engine”** do not appear.

Vector search works by converting words, phrases, or documents into numbers called **vectors**.

For example:

$$ \text{Jet Engine} \rightarrow [0.8,\ 0.2,\ 0.9] $$
$$ \text{Turbine} \rightarrow [0.75,\ 0.25,\ 0.85] $$

$$ \text{Banana} \rightarrow [0.1,\ 0.9,\ 0.2] $$
These vectors represent the **meaning** of the words.

To compare two meanings, vector search often uses **cosine similarity**:

$$ \text{Cosine Similarity} = \frac{A \cdot B}{\|A\| \|B\|}$$

Where:

* $ A \quad \text{is the vector for the search query} $
* $ B  \quad \text{is the vector for a document or word} $
* $ A \cdot B   \quad \text{means the dot product} $
* $|A| \quad \text{and} \quad |B|  \quad \text{are the lengths of the vectors} $

The result is usually between **0 and 1**:

* Close to **1** means very similar
* Close to **0** means not similar

For example:

$$ \text{similarity}(\text{Jet Engine}, \text{Turbine}) = 0.92 $$

$$ \text{similarity}(\text{Jet Engine}, \text{Aircraft Engine}) = 0.95 $$

$$ \text{similarity}(\text{Jet Engine}, \text{Banana}) = 0.12 $$

So vector search can understand that:

$$ \text{Jet Engine} \approx \text{Turbine} \approx \text{Aircraft Engine} $$

because their vectors are close together in meaning.

***

### 4.4.3 Simple Comparison

Normal search asks:

> “Do the exact words match?”

Vector search asks:

> “Are the meanings similar?”

So when searching for:

> “Jet Engine”

Normal search may only find documents containing the exact words **Jet** and **Engine**.

Vector search can also find documents about:

* Turbines
* Propulsion
* Thrust
* Aircraft engines

because these ideas are mathematically close in vector space.

## 4.6 Why Do We Need Chunking in a Vector Database?

Vector databases store **embeddings**—numerical representations of text, images, or other data—to enable semantic search and retrieval.  
**Chunking** is the process of splitting large documents into smaller, semantically meaningful pieces before embedding them.

---

### Why Chunking Is Necessary

### 4.6.1 Embedding Models Have Context Limits
Most embedding models accept a maximum number of tokens (e.g., 512–8,192 tokens).  
If a document length  $L$ exceeds the model limit $C$:


$$ L > C \;\Rightarrow\; \text{Information is truncated or rejected} $$


Chunking ensures all information is embedded and retrievable.

---

### 4.6.2 Better Semantic Precision
Embeddings represent *average semantic meaning* across their input.

If a chunk contains multiple topics, its embedding becomes a semantic mixture:


$$ \vec{e}_{chunk} \approx \frac{1}{n} \sum_{i=1}^{n} \vec{e}_{topic_i} $$


This makes similarity search less precise. Smaller, topic-focused chunks preserve semantic clarity.

---

### 4.6.3 Higher Recall & Better Retrieval
Without chunking, a relevant paragraph buried in a long document may not rank highly during retrieval.  
Chunking increases the chances that at least one relevant piece has **high cosine similarity** with the query:


$$ \cos(\theta) = \frac{\vec{q} \cdot \vec{c}}{\|\vec{q}\|\|\vec{c}\|} $$

---

### What Happens If You Don’t Chunk?

| Problem | Consequence |
|------|-------------|
| Token limit exceeded | Content lost |
| Mixed topics | Poor similarity ranking |
| Large embeddings | Slower search |
| Low recall | Retrieval misses key facts |

In extreme cases, one vector represents an entire document, making fine-grained retrieval impossible.

---

## Summary

- Chunking turns large documents into **searchable semantic units**
- It improves **precision, recall, and system robustness**
- Choose chunk sizes that align with **model limits** and **semantic coherence**

> Good chunking is not just a preprocessing step—it’s a core design decision in vector-based retrieval systems.tokens

`````python
import fitz  # PyMuPDF for reading PDFs
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma
from langchain_core.documents import Document

# --- 1. THE ARCHITECTURE (The Setup) ---
pdf_path = "F18-ABCD-000.pdf"
persist_directory = "./chroma_db"  # Folder where our "digital brain" lives

# The Translator (Embeddings): Converts human language into math (vectors) 
# so the computer can understand the "meaning" of sentences.
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")

# The Slicer (Text Splitter): Breaks long pages into 1000-character chunks.
# The 100-character "overlap" ensures sentences don't get cut in half.
text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=100)

# --- 2. THE FACTORY LINE (Extract & Process) ---
doc = fitz.open(pdf_path)
print(f"Opened PDF with {len(doc)} pages\n")

all_page_documents = []

for i in range(len(doc)):
    page = doc[i]
    text = page.get_text("text")

    if text.strip():
        # Chop the page into manageable, bite-sized pieces
        page_chunks = text_splitter.split_text(text)
        
        # Wrap each piece in a "Document" object with a digital 'sticky note' (metadata)
        # telling us exactly which page it came from.
        for chunk in page_chunks:
            new_doc = Document(
                page_content=chunk,
                metadata={"page": i + 1, "source": pdf_path}
            )
            all_page_documents.append(new_doc)
    else:
        print(f"Skipping page {i+1}: Likely an image or empty page.")

doc.close()

# --- 3. THE ARCHIVE (Creating the Vector Database) ---
# This is the heavy lifting: turning text into math and saving it to disk.
print(f"Encoding {len(all_page_documents)} chunks into the database...")



vector_db = Chroma.from_documents(
    documents=all_page_documents,
    embedding=embeddings,
    persist_directory=persist_directory
)

print("Success! Your searchable digital brain is saved in './chroma_db'.")
`````

## 4.6 🔍 Retrieve

Now that our document chunks are stored, we can search for the most relevant ones.

When the user asks a question, the system looks inside the Vector Database and finds the chunks that are most similar to the question.

---

### 4.6.1 . 📌 What Happens in This Step?

Instead of giving the AI the whole document, we only give it the best matching parts.

These retrieved chunks act like **open-book notes** for the AI.

The AI then uses these notes to write a more accurate answer.

---

### 4.6.2 . 🧠 Simple Example

If the user asks:

> “What is the g-limit above 44,000 pounds?”

The system searches the documents and retrieves the chunk that talks about:

- Gross weight
- G-limiter
- 44,000 pounds
- 5.5 g limit

Then the AI uses that chunk to answer the question.

---

### 💡 Key Idea

A normal chatbot may answer from memory.

A RAG chatbot first retrieves relevant notes, then answers based on those notes.

This helps reduce guessing and makes the answer more factual.

`````python
# --- 1. THE SEARCH QUERY ---
# Define the specific technical question we want to answer.
query = """
        For the F/A-18 Hornet (legacy A/B/C/D variants), below what aircraft gross weight does the Flight Control System (FCS) 
        g-limiter permit a maximum commanded symmetric positive load factor (NzREF) of 7.5 g?
        """

# --- 2. VECTOR SEARCH (Finding Context) ---
# Search the database for the 3 most relevant text snippets (k=3).
# Instead of matching keywords, it matches the "meaning" of the query.
docs = vector_db.similarity_search(query, k=3)

# --- 3. RESULTS INSPECTION ---
# Loop through the matches to verify the information found in the PDF.
for d in docs:
    print(f"--- Result Found ---")
    # Display the actual text found in the document
    print(f"Content: {d.page_content}...")
    
    # Display the source metadata (e.g., the specific page number)
    print(f"Source: Page {d.metadata['page']}\n")
`````

But the basic workflow is the same:

```text
Raw data
   ↓
Split into chunks
   ↓
Convert chunks into embeddings
   ↓
Store embeddings in a vector database
   ↓
Search for relevant chunks
   ↓
Use retrieved chunks as context for an answer
```

`````python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma
from langchain_core.documents import Document

# ---------------------------------------------------------
# 1. THE SETUP
# ---------------------------------------------------------

persist_directory = "./student_chroma_db"

# The Embedding Model:
# This converts text into vectors so the computer can compare meaning.
embeddings = HuggingFaceEmbeddings(
    model_name="all-MiniLM-L6-v2"
)

# Try changing these values!
# Smaller chunk_size = more chunks
# Larger chunk_overlap = more repeated context between chunks
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=50
)

# ---------------------------------------------------------
# 2. RANDOM DATA ENTRIES
# ---------------------------------------------------------
# Instead of reading a PDF, we create some fake data manually.

random_data_entries = [
    {
        "title": "Introduction to RAG",
        "category": "AI",
        "text": """
        Retrieval-Augmented Generation, also called RAG, is a technique that combines search with text generation.
        Instead of asking a language model to answer only from memory, RAG first retrieves relevant information from
        a knowledge base. The retrieved information is then passed to the language model as context. This helps the
        model produce answers that are more accurate, grounded, and specific to the provided data.
        """
    },
    {
        "title": "Vector Databases",
        "category": "Databases",
        "text": """
        A vector database stores information as numerical vectors. These vectors represent the meaning of text,
        images, or other data. When a user asks a question, the question is also converted into a vector.
        The database then searches for stored vectors that are most similar to the query vector. This is called
        similarity search.
        """
    },
    {
        "title": "Chunking Strategy",
        "category": "Text Processing",
        "text": """
        Chunking is the process of splitting long documents into smaller pieces. This is important because language
        models and embedding models have context limits. If chunks are too large, they may contain too much unrelated
        information. If chunks are too small, they may lose important context. Chunk overlap helps preserve meaning
        across chunk boundaries.
        """
    },
    {
        "title": "Embeddings",
        "category": "AI",
        "text": """
        Embeddings are numerical representations of text. Words, sentences, and paragraphs can be converted into
        vectors using an embedding model. Text with similar meaning will usually have similar vectors. For example,
        the sentences 'How do I reset my password?' and 'I forgot my login password' may be close together in
        vector space.
        """
    },
    {
        "title": "Metadata", 
        "category": "RAG Design",
        "text": """
        Metadata is extra information attached to a document or chunk. For example, metadata can include the title,
        source, page number, category, author, or date. Metadata helps us understand where retrieved information
        came from. It can also be used to filter search results.
        """
    }
]

# ---------------------------------------------------------
# 3. SPLIT DATA INTO CHUNKS
# ---------------------------------------------------------

all_documents = []

for entry_id, entry in enumerate(random_data_entries, start=1):
    chunks = text_splitter.split_text(entry["text"])

    for chunk_id, chunk in enumerate(chunks, start=1):
        doc = Document(
            page_content=chunk,
            metadata={
                "entry_id": entry_id,
                "chunk_id": chunk_id,
                "title": entry["title"],
                "category": entry["category"]
            }
        )
        all_documents.append(doc)

print(f"Created {len(all_documents)} chunks.\n")

# Inspect the chunks before storing them
for doc in all_documents:
    print("---- Chunk ----")
    print(f"Title: {doc.metadata['title']}")
    print(f"Category: {doc.metadata['category']}")
    print(f"Chunk ID: {doc.metadata['chunk_id']}")
    print(doc.page_content)
    print()

# ---------------------------------------------------------
# 4. CREATE THE VECTOR DATABASE
# ---------------------------------------------------------

vector_db = Chroma.from_documents(
    documents=all_documents,
    embedding=embeddings,
    persist_directory=persist_directory
)

print("Success! Your student RAG database is saved in './student_chroma_db'.")

# ---------------------------------------------------------
# 5. ASK A QUESTION
# ---------------------------------------------------------

query = """
What is chunk overlap and why is it useful in RAG?
"""

docs = vector_db.similarity_search(query, k=3)

# ---------------------------------------------------------
# 6. INSPECT THE RESULTS
# ---------------------------------------------------------

print("\nSearch Results:\n")

for d in docs:
    print("--- Result Found ---")
    print(f"Title: {d.metadata['title']}")
    print(f"Category: {d.metadata['category']}")
    print(f"Entry ID: {d.metadata['entry_id']}")
    print(f"Chunk ID: {d.metadata['chunk_id']}")
    print()
    print(d.page_content)
    print()
`````
