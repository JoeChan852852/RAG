&nbsp;
<div style="background-color:#FDFBF6; padding:24px; border:1px solid #C9A646; border-radius:6px; font-family:Georgia; box-shadow: 2px 2px 10px rgba(26, 42, 68, 0.05);"> 
    <h2 style="color:#5A1F1F;font-family:Georgia;">
        1.1 🤖 What Is a “Normal” Chatbot?
    </h2>
    <p style="color:#2E2E2E;font-family:Georgia;font-size:17px;line-height:1.7;">
        Trained on a vast breadth of text, conventional AI is remarkably capable, <br>
        yet certain constraints persist:
    </p>
    <ul style="color:#2E2E2E;font-family:Georgia;font-size:17px;line-height:1.8;">
      <li>It remains largely <b>unacquainted with recent events</b></li>
      <li>It lacks access to your <b>private notes and personal documents</b></li>
      <li>It may provide an answer even when <b>lacking certainty</b></li>
      <li>It is occasionally prone to <b>hallucinate</b></li>
    </ul>
    <div style="background-color:#1C39BB;padding:12px;border-left:4px solid #F5BD02;border-radius:6px;font-family:Georgia;">
        <strong style="color:#F5BD02;">💡</strong><br>
        <span style="color:#F5BD02;font-size:16px;line-height:1.7;">
            <b>Hallucinate</b> is to provide an answer with utter conviction that is, in fact, erroneous.Think of a conventional 
            chatbot as a student who has studied extensively, yet is <b>forbidden from consulting their notes</b> during the examination.
        </span>
    </div>
</div>
&nbsp;

<div style="background-color:#FDFBF6; padding:24px; border:1px solid #C9A646; border-radius:6px; font-family:Georgia; box-shadow: 2px 2px 10px rgba(26, 42, 68, 0.05);"> 
    <h2 style="color:#5A1F1F;font-family:Georgia;">
        1.2 🔍 What Is a RAG Chatbot?
    </h2>
    <p style="color:#2E2E2E;font-family:Georgia;font-size:17px;line-height:1.7;">
        <b>RAG</b> denotes <b>Retrieval-Augmented Generation</b>.<br>
        In essence, a RAG-enabled chatbot sits an <b>open-book examination</b>.<br>
        Rather than relying purely on its internal training, the system first <b>consults your documents</b> to extract pertinent 
        details, ensuring its response is grounded in provided evidence.
    </p>
</div>
&nbsp;

<div style="background-color:#FDFBF6; padding:24px; border:1px solid #C9A646; border-radius:6px; font-family:Georgia; box-shadow: 2px 2px 10px rgba(26, 42, 68, 0.05);">

  </h2>
  
  <p style="color:#2E2E2E; font-size:17px; line-height:1.7; margin-bottom:20px;">
    A RAG chatbot follows three simple steps:
  </p>
  
  <ul style="list-style-type:none; padding-left:0; margin:0;">
    <li style="margin-bottom:16px; color:#2E2E2E; font-size:16px; line-height:1.7; display:flex; align-items:flex-start;">

        <strong style="color:#5A1F1F;">Retrieve:</strong><br>
        The chatbot searches your documents and finds the most relevant parts.

    </li>
    
<li style="margin-bottom:16px; color:#2E2E2E; font-size:16px; line-height:1.7; display:flex; align-items:flex-start;">

<h3 style="color:#1A2A44; font-size: 19px; margin-top: 25px;">Visualizing the Logic Flow</h3>

```mermaid
flowchart LR
    A[Start] --> B[1. Import & Setup<br>OpenAI Client with OpenRouter]
    B --> C[2. Define Question<br>and AeroBot System Prompt]
    C --> D[3. Send Request to Model]
    D --> E[4. Receive Response and Print Response]
    E --> G[End]
```

````python
from openai import OpenAI
def LLM(question):
    # --- 1. THE CONNECTION (Setup) ---
    # We use the OpenAI library but point the 'base_url' to OpenRouter.
    # This allows us to access dozens of different AI models with one piece of code.
    api_key = "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" # <---REPLACE WITH YOU API KEY!  
    client = OpenAI(
        base_url="https://openrouter.ai/api/v1",
        api_key=api_key,
    )
  
    # --- 2. THE PERSONA (The Expert) ---
    # This "System" message sets the AI's professional boundaries and tone.
    # It ensures the AI talks like an engineer, not a chatbot.

    role = """
        You are AeroBot – a no-nonsense aerospace & aviation expert.

        Rules:
        - Answer short, direct and to the point
        - Use as few words as possible
        - Never explain unless explicitly asked "explain" or "why"
        - No introductions, no conclusions, no apologies, no filler phrases
        - Only talk about aerospace, aircraft, airlines, aerodynamics, propulsion, avionics, 
                spaceflight, airports, pilots, incidents – nothing else
        - If question is unrelated → reply only: "Off-topic. Aerospace/plane question only."

        Examples of good answers:
        User: What is the cruise speed of A320?
        You: Mach 0.78 (~828 km/h)

        User: Why do 737 MAX have those split wingtips?
        You: Increases aspect ratio, reduces induced drag, improves fuel efficiency ~3-4%

        User: Tell me about yourself
        You: Off-topic. Aerospace/plane question only.
        """

    # --- 3. THE EXECUTION (The API Call) ---
    # We package everything up and send it to the model.
    # The 'system' role is the personality; the 'user' role is the specific task.
    response = client.chat.completions.create(
            # Using a free model for testing
            model="openrouter/free",
            messages=[
                {"role": "system", "content": role},
                {"role": "user", "content": question}
                ]
            )
        
    # --- 4. THE DELIVERY (Display Results) ---
    # Pull the text content out of the AI's response package and print it.
    print("🤖 AI Says:")
    print(response.choices[0].message.content)
````

````python
question = """
For the F/A-18 Hornet, below what aircraft gross weight does the Flight Control System (FCS)
g-limiter permit a maximum commanded symmetric positive load factor (NzREF) of 7.5 g?understand
"""
LLM(question)
````

    <h2 style="color:#5A1F1F;font-family:Georgia;border-bottom: 1px double #C9A646; padding-bottom: 10px;">

    </h2>
    <p style="color:#2E2E2E;font-family:Georgia;font-size:17px;line-height:1.7;">
        Before we explore Retrieval-Augmented Generation (RAG), let us test the most intuitive method:

    <h3 style="color:#1A2A44; font-size: 20px; margin-top: 35px; border-left: 4px solid #C9A646; padding-left: 15px;">

    </h3>
    <p style="color:#2E2E2E; font-size:16px; line-height:1.6;">
        Prompt stuffing involves feeding a vast quantity of text directly into the AI, requesting a response based solely on that material.
    </p>
    <div style="background-color: #FFF; border: 1px dashed #C9A646; padding: 15px; margin: 15px 0;">
        <strong style="color:#5A1F1F; text-transform: uppercase; font-size: 13px; letter-spacing: 1px;">The Workflow:</strong>
        <ol style="color:#2E2E2E; line-height: 1.8; margin-top: 10px;">
            <li>Copy a lengthy document.</li>
            <li>Paste the text into the chat.</li>
            <li>Append your question at the end.</li>
            <li>Command the AI to provide an answer.</li>
        </ol>
    </div>

    </p>

````python
import fitz  # PyMuPDF: The library used to "read" and parse PDF files

# The file path is the specific location of your technical manual
pdf_path = "F18-ABCD-000.pdf"

# --- 1. ACCESS THE DOCUMENT ---
# Open the file so we can begin scanning its contents
doc = fitz.open(pdf_path)
print(f"Success! Opened a PDF with {len(doc)} pages\n")

# --- 2. INITIALIZE THE "COLLECTION" ---
# This variable will hold all the text we extract from every page.
# Think of it as a singl@hkuenggailt.onaliyun.come, long digital transcript.
PDF_content = ""

# --- 3. THE EXTRACTION LOOP ---
# We visit every page one-by-one, from the first to the last.
for i in range(len(doc)):
    page = doc[i]  # "Pick up" the current page
    
    # OCR/Text Scanning: Convert the visual layout into readable strings.
    text = page.get_text("text") 

    # --- 4. FORMATTING THE TRANSCRIPT ---
    # We add visual dividers (═) so the AI (and we) can see where pages start and end.
    # This acts like a "table of contents" built directly into the text.
    PDF_content += f"\n{'═' * 80}\n"
    PDF_content += f"Page {i+1:3d} of {len(doc)}\n"
    PDF_content += f"{'═' * 80}\n\n"

    # --- 5. VALIDATION ---
    # We check if the page actually contains text.
    # If a page is just a photo or diagram, the "text" will be empty.
    if text.strip():
        PDF_content += text + "\n\n"
    else:
        PDF_content += "[No extractable text on this page - likely an image or diagram]\n\n"

# --- 6. FINALIZING ---
# Close the file to free up system memory.
doc.close()

# The 'PDF_content' variable is now ready to be sent to an AI or a database!
print("Extraction complete. Text is ready for processing.")
````

````python
from openai import OpenAI

# --- 1. THE CONNECTION (Setup) ---
# We use the OpenAI library but point the 'base_url' to OpenRouter.
# This allows us to access dozens of different AI models with one piece of code.
api_key = "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"
client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=api_key,
    )


# --- 2. THE TARGET (The Question) ---
# This is the specific mystery we are trying to solve using the technical manual.
# We are asking about the F/A-18 Hornet's "G-limiter." 
# Think of a G-limiter as the car's speed limiter, but for how hard a plane can turn.
question = """
For the F/A-18 Hornet (legacy A/B/C/D variants), below what aircraft gross weight does the 
Flight Control System (FCS) g-limiter permit a maximum commanded symmetric positive 
load factor (NzREF) of 7.5 g?
"""

# --- 3. THE TEMPLATE (The "Open-:Book Exam") ---
# We use an 'f-string' (the f before the quotes) to plug in our PDF text.
# This tells the AI: "Here is a massive pile of data (PDF). Look ONLY here."
prompt = f"""
You are a precise technical assistant. Answer ONLY using the provided CONTEXT.

QUESTION:
{question}

CONTEXT:
{PDF_content}

Rules — you MUST follow all of them:
1. Use ONLY facts explicitly stated in the CONTEXT. (No guessing!)
2. Do NOT use outside knowledge. (Ignore what you learned on the internet.)
3. If the answer isn't in the text, say: "Not enough information."
4. Keep it short, factual, and use bullet points.
"""

# --- 4. THE PERSONA (The Expert) ---
# This "System" message sets the AI's professional boundaries and tone.
# It ensures the AI talks like an engineer, not a chatbot.
role = """
You are AeroBot – a no-nonsense aerospace & aviation expert.

Rules:
- Answer short, direct and to the point
- Use as few words as possible
- Never explain unless explicitly asked "explain" or "why"
- No introductions, no conclusions, no apologies, no filler phrases
- Only talk about aerospace, aircraft, airlines, aerodynamics, propulsion, avionics, spaceflight, airports, pilots, incidents – nothing else
- If question is unrelated → reply only: "Off-topic. Aerospace/plane question only."

Examples of good answers:
User: What is the cruise speed of A320?
You: Mach 0.78 (~828 km/h)

User: Why do 737 MAX have those split wingtips?
You: Increases aspect ratio, reduces induced drag, improves fuel efficiency ~3-4%

User: Tell me about yourself
You: Off-topic. Aerospace/plane question only.
"""

# --- 5. THE EXECUTION (The API Call) ---
# We package everything up and send it to the model.
# The 'system' role is the personality; the 'user' role is the specific task.
response = client.chat.completions.create(
        # Using a free model for testing
        model="openrouter/free",
        messages=[
            {"role": "system", "content": role},
            {"role": "user", "content": prompt}
            ]
        )
        
# --- 6. THE DELIVERY (Display Results) ---
# Pull the text content out of the AI's response package and print it.
print("🤖 AI Says:")
print(response.choices[0].message.content)
````
